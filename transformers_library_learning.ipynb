```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

reviews = ["This film was a waste of time","This film was very nice"]

results = classifier(reviews)

print(results)
```

```python
from transformers import pipeline

qa = pipeline("question-answering")

contxt = "The Transformers library by Hugging Face provides state-of-the-art NLP models. It supports tasks like translation, summarization, and question answering."

qus = "What does the Transformers library support?"

result = qa(question=qus, context=contxt)

print(result)
```

```python
print(result['answer'])
```

```python
from transformers import pipeline

ner = pipeline("ner", grouped_entities = True)

text = "Elon Musk founded spaceX in 2002 and Tesla Motors in 2003"

entities = ner(text)

print(entities)

```

```python
for entity in entities:
  print(f"{entity['word']} | {entity['entity_group']}")
```

```python
from transformers import pipeline

pos_tagger = pipeline(model="vblagoje/bert-english-uncased-finetuned-pos")

text = "Elon Musk founded spaceX in 2002 and Tesla Motors in 2003"

pos_tags = pos_tagger(text)

print(pos_tags)
```

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

text = "The new iPhone is released with a better camera and improved battery life"

candidate_labels = ["Technology","Health","Politics","Sports"]

result = classifier(text, candidate_labels)

print(result)
```
